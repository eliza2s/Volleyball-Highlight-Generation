##  SPLITTING VIDEO INTO FRAMES

```python
import cv2
import os

video_path = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\NEW_HIGH\watch.mp4'
output_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\NEW_HIGH\watch_frames_2min'
os.makedirs(output_folder, exist_ok=True)

cap = cv2.VideoCapture(video_path)
if not cap.isOpened():
    print("❌ Could not open video")
    exit()

fps = int(cap.get(cv2.CAP_PROP_FPS))
print(f"Video FPS: {fps}")
max_frames = fps * 120  # 2 minutes

frame_count = 0
while cap.isOpened() and frame_count < max_frames:
    ret, frame = cap.read()
    if not ret:
        print("⚠️ No more frames to read or error reading frame.")
        break

    # Save as frame_0000.png, frame_0001.png, ...
    filename = os.path.join(output_folder, f"frame_{frame_count:04d}.png")
    cv2.imwrite(filename, frame)

    if frame_count % 100 == 0:
        print(f"✅ Saved frame {frame_count}")

    frame_count += 1

cap.release()
print(f"✅ Done! Extracted {frame_count} frames to: {output_folder}")
```

## VISUALIZATION OF PEOPLE CLASSIFIED FRAMES INTO VIDEO

```python
import torch
import cv2
import os

# === 1. Load YOLOv5 Model ===
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)
model.conf = 0.25  # Confidence threshold

# ✅ Target class for people
TARGET_CLASS_NAME = 'person'

# === 2. Paths ===
frames_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\NEW_HIGH\watch_frames_2min'
output_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\NEW_HIGH\watch_frames_person_detected'
os.makedirs(output_folder, exist_ok=True)

# === 3. Frame limit ===
FPS = 30
DURATION_SECONDS = 120  # 2 minutes
MAX_FRAMES = FPS * DURATION_SECONDS  # 3600

# === 4. Load and sort frames ===
frame_files = sorted([f for f in os.listdir(frames_folder) if f.lower().endswith(('.jpg', '.png'))])
frame_files = frame_files[:MAX_FRAMES]

print(f"Processing up to {len(frame_files)} frames...")

# === 5. Run detection and draw bounding boxes ===
for idx, filename in enumerate(frame_files):
    frame_path = os.path.join(frames_folder, filename)
    image = cv2.imread(frame_path)

    if image is None:
        print(f"⚠️ Skipped invalid image: {filename}")
        continue

    # Inference
    results = model(image)
    detections = results.pandas().xyxy[0]

    # ✅ Filter only for 'person'
    person_detections = detections[detections['name'] == TARGET_CLASS_NAME]

    # Draw bounding boxes
    for _, row in person_detections.iterrows():
        xmin, ymin, xmax, ymax = map(int, [row['xmin'], row['ymin'], row['xmax'], row['ymax']])
        cv2.rectangle(image, (xmin, ymin), (xmax, ymax), (255, 0, 0), 2)  # Blue box

    # Save annotated image
    output_path = os.path.join(output_folder, f"frame_{idx:04d}.jpg")
    cv2.imwrite(output_path, image)
    print(f"✅ Saved {output_path}")

print("🎉 Done! All frames processed with person bounding boxes.")
```

```python
import cv2
import os

# === 1. Input Folder ===
# Replace this with your latest output folder if different
detected_frames_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\NEW_HIGH\watch_frames_person_detected'

# === 2. Output Video Path ===
output_video_path = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\NEW_HIGH\person_detection_output.mp4'

# === 3. Frame Settings ===
fps = 30
frame_files = sorted([
    f for f in os.listdir(detected_frames_folder)
    if f.lower().endswith(('.jpg', '.png'))
])

# === 4. Get frame size from the first frame ===
first_frame_path = os.path.join(detected_frames_folder, frame_files[0])
first_frame = cv2.imread(first_frame_path)
height, width, _ = first_frame.shape

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
video_writer = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

# === 5. Write frames to video and display ===
for filename in frame_files:
    frame_path = os.path.join(detected_frames_folder, filename)
    frame = cv2.imread(frame_path)

    if frame is None:
        print(f"⚠️ Could not read frame: {filename}")
        continue

    # Show video while writing
    cv2.imshow("Detection Playback", frame)
    video_writer.write(frame)

    key = cv2.waitKey(int(1000 / fps))
    if key == 27:  # ESC key to stop early
        break

video_writer.release()
cv2.destroyAllWindows()

print(f"🎥 Video saved to: {output_video_path}")
```
