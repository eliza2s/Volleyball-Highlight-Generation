## CLASSIFICATION OF THE FRAMES AND SAVING IN A FOLDER


```python
import torch
import cv2
import os

# Load YOLOv5 model
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)

# Input/output folders
frames_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\archive (1)\Volley Ball Match'
output_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\archive (1)\output' 
os.makedirs(output_folder, exist_ok=True)

# Limit to 5400 frames (3 minutes at 30 fps)
MAX_FRAMES = 5400

# Get sorted list of frame files
frame_files = sorted([f for f in os.listdir(frames_folder) if f.endswith(('.png', '.jpg', '.jpeg'))])
frame_files = frame_files[:MAX_FRAMES]  # Limit to 3 minutes

# Processing loop
for idx, filename in enumerate(frame_files):
    frame_path = os.path.join(frames_folder, filename)
    image = cv2.imread(frame_path)

    if image is None:
        print(f"Skipped {filename}: Not a valid image")
        continue

    # Run inference
    results = model(image)

    # Render detections
    results.render()
    annotated_image = results.ims[0]

    # Save result
    output_path = os.path.join(output_folder, f"frame_{idx:04d}.jpg")
    cv2.imwrite(output_path, annotated_image)
    print(f"Processed {idx+1}/{len(frame_files)}: {filename}")
```

## VISUALIZATION OF CLASSIFICATION IN A VIDEO

```python
import cv2
import os

# Folder containing annotated frames
input_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\archive (1)\output'
output_video_path = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\archive (1)\highlight_output_2min.mp4'

# Get sorted frame filenames
frame_files = sorted([f for f in os.listdir(input_folder) if f.endswith(('.jpg', '.png'))])

# Limit to first 3600 frames (2 minutes at 30 FPS)
frame_files = frame_files[:3600]

# Read first frame to get video dimensions
first_frame = cv2.imread(os.path.join(input_folder, frame_files[0]))
height, width, _ = first_frame.shape

# Create video writer
fps = 30
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

# Write frames to video
for frame_name in frame_files:
    frame_path = os.path.join(input_folder, frame_name)
    frame = cv2.imread(frame_path)
    if frame is not None:
        out.write(frame)
    else:
        print(f"Skipped frame: {frame_name}")

out.release()
print(f"✅ 2-minute video saved at: {output_video_path}")
```

## Ball-only bounding box frames

```python
import torch
import cv2
import os

# === 1. Load YOLOv5 Model ===
# Load the pre-trained YOLOv5s model from ultralytics
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)
model.conf = 0.25  # Confidence threshold (can adjust)

# Target class (COCO class 37 = sports ball)
TARGET_CLASS_NAME = 'sports ball'

# === 2. Setup Paths ===
# Input folder containing frames like frame_0000.jpg, frame_0001.jpg, ...
frames_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\archive (1)\Volley Ball Match'

# Output folder for annotated images
output_folder = r'C:\Users\Nitro v15\Desktop\Turboline hackathone\archive (1)\output_ball_only'
os.makedirs(output_folder, exist_ok=True)

# === 3. Limit to 2 Minutes Worth of Frames ===
FPS = 30
DURATION_SECONDS = 120  # 2 minutes
MAX_FRAMES = FPS * DURATION_SECONDS  # 3600

# === 4. Load and Sort Frame Files ===
frame_files = sorted([f for f in os.listdir(frames_folder) if f.lower().endswith(('.jpg', '.png'))])
frame_files = frame_files[:MAX_FRAMES]

print(f"Processing up to {len(frame_files)} frames...")

# === 5. Loop through Frames and Annotate Only Balls ===
for idx, filename in enumerate(frame_files):
    frame_path = os.path.join(frames_folder, filename)
    image = cv2.imread(frame_path)

    if image is None:
        print(f"⚠️ Skipped invalid image: {filename}")
        continue

    # Run YOLO inference
    results = model(image)

    # Get detection results as a pandas DataFrame
    detections = results.pandas().xyxy[0]

    # Filter only 'sports ball' detections
    ball_detections = detections[detections['name'] == TARGET_CLASS_NAME]

    # Draw bounding boxes manually
    for _, row in ball_detections.iterrows():
        xmin, ymin, xmax, ymax = map(int, [row['xmin'], row['ymin'], row['xmax'], row['ymax']])
        confidence = row['confidence']
        label = f"{TARGET_CLASS_NAME} {confidence:.2f}"
        # Draw rectangle
        cv2.rectangle(image, (xmin, ymin), (xmax, ymax), (0, 255, 0), 2)
        # Draw label
        cv2.putText(image, label, (xmin, ymin - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    # Save the output frame
    output_path = os.path.join(output_folder, f"frame_{idx:04d}.jpg")
    cv2.imwrite(output_path, image)
    print(f"✅ Saved {output_path}")

print("Done! Ball-only bounding box frames saved.")
```
